# Descargar el dataset desde Google Drive

En este notebook se descarga el dataset desde Google Drive. Para ello, se utiliza la librería `gdown` que permite descargar archivos desde Google Drive.

```python
import gdown

# ID del archivo de Google Drive
file_id = '1L0z2xVmkoEMlSmZ1jR5MGdNkb4WU6Mek'

# Nombre del conjunto de datos
dataset_name = 'yolo_signals_cbba_with_augmented_data'

# Destination path
destination = f'{dataset_name}.zip'

# URL de descarga
url = f'https://drive.google.com/uc?id={file_id}'

# Descargar el archivo
gdown.download(url, destination, quiet=False)  
```

```python
import zipfile
with zipfile.ZipFile(destination, 'r') as zip_ref:
    zip_ref.extractall(dataset_name)
```

```python
# Mover el dataset a la carpeta correspondiente
import shutil
shutil.move(dataset_name, '../dataset/processed')
```

```python
# Eliminar el archivo .zip
import os
os.remove(destination)
```

```python
# Verificar que el archivo fue descargado correctamente
import os
os.listdir(f'../dataset/processed/{dataset_name}')
```
