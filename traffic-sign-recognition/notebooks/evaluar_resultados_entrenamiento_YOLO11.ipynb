```python
import pandas as pd
```

```python
# Cargar nuevamente el archivo CSV
file_path = "../resultados/YOLO11/300_epochs/results.csv"
df = pd.read_csv(file_path)

# Limpiar los nombres de las columnas eliminando espacios extra
df.columns = df.columns.str.strip()

# Mostrar las primeras filas para verificar los datos
df.head()
```

```python
import matplotlib.pyplot as plt

# Graficar la evolución de las métricas de entrenamiento y validación

plt.figure(figsize=(18, 8))

# Pérdidas de entrenamiento y validación
plt.subplot(1, 2, 1)
plt.plot(df["epoch"], df["train/box_loss"], label="train/box_loss")
plt.plot(df["epoch"], df["train/cls_loss"], label="train/cls_loss")
plt.plot(df["epoch"], df["train/dfl_loss"], label="train/dfl_loss")
plt.xlabel("Época")
plt.ylabel("Pérdida")
plt.title("Evolución de la pérdida - Entrenamiento")
plt.axvline(x=102, color='red', linestyle='--', label="Época 102")
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(df["epoch"], df["val/box_loss"], label="val/box_loss")
plt.plot(df["epoch"], df["val/cls_loss"], label="val/cls_loss")
plt.plot(df["epoch"], df["val/dfl_loss"], label="val/dfl_loss")
plt.xlabel("Época")
plt.ylabel("Pérdida")
plt.title("Evolución de la pérdida - Validación")
plt.axvline(x=102, color='red', linestyle='--', label="Época 102")
plt.legend()



plt.show()

# Graficar la evolución de precisión, recall y mAP
plt.figure(figsize=(18, 8))

plt.subplot(1, 3, 1)
plt.plot(df["epoch"], df["metrics/precision(B)"], label="Precisión")
plt.xlabel("Época")
plt.ylabel("Precisión")
plt.title("Evolución de la Precisión")
plt.axvline(x=102, color='red', linestyle='--', label="Época 102")

plt.legend()

plt.subplot(1, 3, 2)
plt.plot(df["epoch"], df["metrics/recall(B)"], label="Recall")
plt.xlabel("Época")
plt.ylabel("Recall")
plt.title("Evolución del Recall")
plt.axvline(x=102, color='red', linestyle='--', label="Época 102")
plt.legend()

plt.subplot(1, 3, 3)
plt.plot(df["epoch"], df["metrics/mAP50(B)"], label="mAP@0.5")
plt.plot(df["epoch"], df["metrics/mAP50-95(B)"], label="mAP@0.5:0.95")
plt.xlabel("Época")
plt.ylabel("mAP")
plt.title("Evolución de mAP")
plt.axvline(x=102, color='red', linestyle='--', label="Época 102")
plt.legend()

plt.show()
```
